```python
import pandas as pd
from mlxtend.frequent_patterns import apriori, association_rules
```

```python
file_path = r"C:\Users\Ksholkkamy\Downloads\customer_churn_realistic.csv.xls"

df = pd.read_csv(file_path)

print(df.head())
print(df.shape)
```

```python
df_encoded = pd.get_dummies(df)

# تحويل القيم لـ True/False
df_encoded = df_encoded.astype(bool)

print(df_encoded.head())
```

```python
frequent_itemsets = apriori(
    df_encoded,
    min_support=0.2,
    use_colnames=True
)

print(frequent_itemsets.head())
```

```python
rules = association_rules(
    frequent_itemsets,
    metric="confidence",
    min_threshold=0.6
)

print(rules.head())
```
